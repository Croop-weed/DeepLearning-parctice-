```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
docs = nlp("I will google about facebook")
```

```python
docs.text
```

```python
for i in docs:
  print(i,"\t","->\t",i.pos_,"\t",i.tag_,"\t",spacy.explain(i.tag_))
```
